# Matplotlib part 1

```python
import matplotlib.pyplot as plt
```

```python
#to see the graph you create inside the jupyter notebooks
```

```python
%matplotlib inline
```

```python
import numpy as np
x = np.linspace(0,5,11)
y = x**2
```

```python
x
```

```python
y
```

```python
# functional
plt.plot(x,y)
#to add the label
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title('Title')
#to get the plot shown
plt.show()
```

### Putting multiple plots on the same canvas

```python
# first argument: number of rows, number of columns, plot
plt.subplot(1, 2, 1)
plt.plot(x,y,'r')


plt.subplot(1,2,2)
plt.plot(y,x,'b')
plt.show()
```

```python
# Object oriented method

fig = plt.figure() #blank canvas

axes = fig.add_axes([0.1,0.1,0.8,0.8])
# passes a list of axes: 2 first are localtion and then 2 follows are size
axes.plot(x,y)
axes.set_xlabel('X label')
axes.set_ylabel('Y label')
axes.set_title('Set title')
plt.show()
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0.1, 0.1, 0.8, 0.8]) #the numbers should be between 0 and 1
axes2= fig.add_axes([0.2,0.5, 0.4, 0.3])
axes1.plot(x,y)
axes1.set_title('LARGER PLOT')
axes2.plot(y,x)
axes2.set_title('SMALLER PLOT')
plt.show()
# basically the fig.add_axes list are the distance from the canvas, the the size of the graph
```

```python
# create an empty figure, which has nothing to show
# then you add axes to it
fig = plt.figure()
axes1 = fig.add_axes([0.1,0.1,0.8,0.8])
axes1.plot(x,y)
plt.show()
```

# Matplotlib part2

```python
fig, axes = plt.subplots()
# the axes here is actually the one we create before axes1, 
# plt.subplots is different from plt.figure is plt.subplots add the figure automatically based on how many rows and cols you want
plt.show()
```

```python
fig, axes = plt.subplots()
axes.plot(x,y)
plt.show()
```

```python
fig, axes = plt.subplots(nrows = 3, ncols =2)
plt.tight_layout #to not have overlapping
plt.show()
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
plt.show()
```

```python
axes 
#an array of matplotlib object
```

```python
fig, axes = plt.subplots(nrows = 1, ncols = 2)

for current_ax in axes:
    current_ax.plot(x,y)

plt.show()
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
# you can pass the value of the figures based on the index of the graph
axes[0].plot(x,y)
axes[0].set_title('First Plot')
axes[1].plot(y,x)
axes[1].set_title('Second Plot')
# to avoid overlapping
plt.tight_layout()
plt.show()
```

# Figure Size and DPI 

```python
fig = plt.figure(figsize=(5,2), dpi=100)
# figsize: first argument vertical size, sec: horizontal size
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y)
plt.show()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=1,figsize = (8,2))
axes[0].plot(x,y)
axes[1].plot(y,x)
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,x**2, label ='x square')
ax.plot(x,x**3, label = 'x cubed')
# to add legend, can specify localtion so that it doesn't overlap
ax.legend(loc = 0)
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,x**2, label ='x square')
ax.plot(x,x**3, label = 'x cubed')
# to add legend, can specify localtion so that it doesn't overlap
ax.legend(loc = 10)
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,x**2, label ='x square')
ax.plot(x,x**3, label = 'x cubed')
# to add legend, can specify localtion so that it doesn't overlap
ax.legend(loc = (0.1,0.1))
plt.show()
```

# Part 3
Setting color, line width, line wide

## Plot Appearance

```python
fig = plt.figure()
ax = fig.add_axes([0,0,0.8,0.8])
ax.plot(x,y, color = 'purple', linewidth=3, alpha = 1, linestyle = '--')
# this also works ax.plot(x,y, color = 'purple', lw=3, alpha = 0.5)
# you can add a RGB Hex Code
# 1 is default linewidth, the bigger the number the more bold the line is
# alpha shows how transparent the line is, should be between 0 and 1
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,0.8,0.8])
ax.plot(x,y, color = 'purple', lw=2, alpha = 1, linestyle = '-.')
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,0.8,0.8])
ax.plot(x,y, color = 'purple', lw=2, alpha = 1, ls = '-')
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,0.8,0.8])
ax.plot(x,y, color = 'purple', lw=1, alpha = 1, marker = 'o', markersize=10,
       markerfacecolor='yellow', markeredgewidth=3, markeredgecolor='green')
plt.show()
```

## Control Y and X axes

```python
fig = plt.figure()
ax = fig.add_axes([0,0,0.8,0.8])
ax.plot(x,y, color = 'purple', lw=2, alpha = 1, ls = '--')
ax.set_xlim([0,1])
# ax.set_xlim() is to pass a lower bound and upper bound limit
ax.set_ylim([0,1])
plt.show()
```
