# Supervised Learning Overview
- Supervised learning algorithms are trained using labeled examples, such as an input where the desired output is known.
- eg: a segment of text could have a category label, such as:
  + spam vs legitimate email
  + positive vs negative movie review
- The network receives a set of inputs along with the corresponding correct outputs
-  and the algorithm learns by comparing its actual output with correct output to find errors
-  it then modifies the model accordingly

- Supervised learning is commonly used in applications where historical data predicts likely future events
  
  

- Data is often split into 3 sets:
  + training data: used to train model parameters
  + validation data: used to determine what model hyperparameters to adjust
  + test data: used to get  some final performance metric

Meaning: after we saw the results on the final test set we don't get to go back and adjust any model parameters.
*This final measure is what we label the true performance of the model to be*

# Evaluating Performance - Classification Error Metrics

### Model Evaluation
- after our machine learning process is complete, we will use performance metrics to evaluate how our model did.
- We'll discuss classification metrics in more detail
  
- The key classification metrics we need to understand are:
  + Accuracy
  + Recall
  + Precision
  + F1-Score

- Accurray: in classification problems is the **number of correct predictions** made by the model,
divided by the **total number of predictions**
  + is uselful when target classes are well balanced
  + eg: in our example: we have roughly same amount of cat images as dog images
  + Accuracy is **not** a good choice with **unbalanced** classes

- Recall: is ability of a model to find relevant cases within a dataset.
  + The precise definition of recall is the number of true positives **divided** by the number of true positives,
    plus the number of **false negatives**.

- Precision: is ability of a classificaion model to identify only relevant data points.
  + Precision is defined as the number of true positives divided by the number of true positives,
    plus the number of **false positives**

- Often you have a trade-off between Recall and Precision.
  + While Recall expresses the ability to find all relevant instances in a dataset,
    precision expresses the proportion of the data points the model says were relevant actually were relevant.

- F1-Score: to find an optimal blend of precision and recall we can combine the two metrics using what is called F1 score.
  + F1-Score = 2*(precision*recall)/(precision+recall)
 
- We can also view all the correctly classified versus incorrectly classified images in the form of a confusion matrix
  
  + Use confusion matrix to compare the predicted values versus true values

# Evaluating Performance Regression

- Regression is a task when a model attempts to predict continuous values (unlike categorical values, which is classification)
- Accuracy or Recall is not helpful for **continuous** values.
- EG: predict the price of a house given its features is a regression task.
- EG: predict the country a house is in given its features is a classification task. 


- The most common evaluation metrics for regression:
  + Mean Absolute Error
  + Mean Squared Error
  + Root Mean Square Error

- Mean Absolute Error (MAE)
  + This is the mean of the absolute value of errors.
  + Easy to understand 
  + MAE won't punish large errors however 

- Mean Squared Error (MSE): is the mean of the squared errors
  + Larger errors are noted more than with MAE, making MSE more popular 

- Root Mean Square Error (RMSE): is the root of the mean of the squared errors.
  + Most popular (has same units as y) 
  + 

- Is this value of RMSE good? --> context is everything --> A RMSE of 10$ is good for predicting price of a house,
but horrible for predicting the price of a candy bar.
- You should compare your error metric to the average value of the label in your dataset to get an intuition of its overall performance.
- Domain knowledge is important.
